```python
import csv
import pandas as pd

#we create a new dataframe and safe it as csv
df_stimuli = pd.DataFrame(columns=['id','sentence','image_file','audio_file','objects','role','timeout',])

with open ('sentences.csv', 'r') as file:
    reader = csv.reader(file)

    #we carry over the timeout set in the original open sesame table just to be safe
    timeout=3000

    indx = 0
    for row in reader:
        print(row)
        
        objects = [row[3],row[4],row[5],row[6]] if len(row) > 6 else [row[3],row[4],row[5]]
        if indx <25:
            sentence=row[1] #taking the restrictive sentence for the first 25
            audio_file = f"{indx}_r"
            role='restrictive'
        else:
            sentence=row[2] #takling the non restictive for the last 25
            audio_file = f"{indx}_n"
            role='non-restrictive'

        imge_file = f"pic{indx+1}.png"#tara saved the images staring at 1 while i saved the audios starting at 0. We can correct this mismatch at the next iteration of stimuli creation
        

        df_stimuli.loc[len(df_stimuli)] = {'id': indx, 'sentence': sentence, 'image_file': imge_file, 'audio_file': audio_file, 'objects': objects, 'role':role, 'timeout': timeout}
        indx +=1

#adding stimuli with mismatch here: (currently only one created: pineapple girl)
df_stimuli.loc[len(df_stimuli)] = {'id': indx, 'sentence': 'The girl will eat the pineapple', 'image_file': 'pic_mismatch_1.png', 'audio_file': 'mismatch_1.wav', 'objects': objects, 'role': 'mismatch', 'timeout': timeout}

df_stimuli
```
